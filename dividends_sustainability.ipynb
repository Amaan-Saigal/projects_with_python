```python
# import libraries
import pandas as pd
import yfinance as yf
import numpy as np
```

```python
# ['XOM','CVX','JNJ','PEP','MRK','COP','MDT','VZ','O','GIS']
# define tickers
tickers = ['XOM','CVX','JNJ','PEP','MRK','COP','MDT','VZ','O','GIS']
# download the data
start_date = "2014-01-01"
end_date = "2025-12-31"
stocks = yf.download(tickers, start=start_date, end=end_date,actions=True, group_by='ticker')
stocks.head()
```

```python
stocks.columns
```

```python
# drop all the null values in stocks DataFrame
stocks = stocks.dropna(how='all')
stocks
```

```python
# make a seperate dataset for revenue,netincome and totaldebt
ticker_statements = []
for y in tickers:
    tickery = yf.Ticker(y)
    revenue = tickery.financials.loc['Total Revenue']
    if revenue is not None and not revenue.empty:
        revenue = revenue.reset_index()
        revenue.columns = ["Date", "Revenue"]
        revenue["Ticker"] = y
        ticker_statements.append(revenue)
    netincome = tickery.financials.loc['Net Income']
    if netincome is not None and not netincome.empty:
        netincome = netincome.reset_index()
        netincome.columns = ["Date", "NetIncome"]
        netincome["Ticker"] = y
        ticker_statements.append(netincome)
    totaldebt = tickery.balance_sheet.loc['Total Debt']
    if totaldebt is not None and not totaldebt.empty:
        totaldebt = totaldebt.reset_index()
        totaldebt.columns = ["Date", "TotalDebt"]
        totaldebt["Ticker"] = y
        ticker_statements.append(totaldebt)
# print("revenue:", revenue)
# print("netincome:", netincome)
# print("totaldebt:", totaldebt)
print(ticker_statements)
```

```python
# make a single dataframe for the fiancial statements
financials_df = pd.concat(ticker_statements, axis=0)
# # make sure the Date column is in datetime format
# financials_df['Date'] = pd.to_datetime(financials_df['Date'])
print(financials_df)
```

```python
#reset index of combined_statements and stocks DataFrames
stocks2 = (stocks.stack(level=0).rename_axis(['Date', 'Ticker']).reset_index())
financials_df2 = financials_df.reset_index()
```

```python
# check index levels of both DataFrames
fdflevel = financials_df2.index.nlevels
slevel = stocks2.index.nlevels
print(fdflevel, slevel)
```

```python
stocks2['Date'] = pd.to_datetime(stocks2['Date'])
financials_df2['Date'] = pd.to_datetime(financials_df2['Date'])

stock_data = pd.merge(stocks2,
                      financials_df2,
                      how='left',
                      on=["Ticker", "Date"])
stock_data
```

```python
# save the datadframe to a CSV file
dividends_sustainability = stock_data.to_csv("dividends_sustainability_data.csv", index=False)
```
